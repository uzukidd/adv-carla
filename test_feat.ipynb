```python
import torch
```

```python
w = torch.rand((1, 5))
w.requires_grad = True
x = torch.rand((5, 3))
x.requires_grad = True
y = torch.matmul(w, x)
print(y)
y_hat = y[:, [0, 2]]
print(y_hat)
ending = y_hat.sum()
ending.backward()
print(x.grad)
current_node = ending
while True:
    if hasattr(current_node, "grad_fn") :
        print(current_node.grad_fn)
        current_node = current_node.grad_fn.next_functions[0][0]
    elif hasattr(current_node, "next_functions") :
        print(current_node.next_functions)
        current_node = current_node.next_functions[0][0]
    else:
        break
    
```
